```python

import pandas as pd

# dataset lode
df = pd.read_csv("train.csv")

```

```python
# Problem 1 – Dataset Inspection.


# 1. Display the first 10 rows.
print("First 10 rows:")
print(df.head(10))
```

```python
# 2. Find the number of rows and columns.
print("\nNumber of rows and columns:")
print(df.shape)
```

```python
# 3. Identify all column names.
print("\nColumn names:")
print(df.columns)
```

```python
# 4. Determine the data type of each column.
print("\nData types:")
print(df.dtypes)
```

```python
# 5. Display descriptive statistics for numerical columns.
print("\nDescriptive statistics:")
print(df.describe())
```

```python
# Problem 2 – Data Filtering. Using the Titanic dataset:

# 1. Display passengers belonging to first class (Pclass == 1).
print("First class passengers:")

first_class = df[df["Pclass"] == 1]

print(first_class)
```

```python
# 2. Find passengers younger than 18.
print("\nPassengers younger than 18:")

young_passengers = df[df["Age"] < 18]

print(young_passengers)
```

```python
# 3. Find male passengers who survived
print("\nMale passengers who survived:")

male_survivors = df[
    (df["Sex"] == "male") &
    (df["Survived"] == 1)
]

print(male_survivors)
```

```python
# 4. Find passengers who paid a fare greater than 100.
print("\nPassengers who paid more than 100:")

high_fare_passengers = df[df["Fare"] > 100]

print(high_fare_passengers)
```

```python
# Problem 3 – Sorting and Aggregation. Using the Titanic dataset:

# 1. Display the 10 passengers who paid the highest fares.
print("Top 10 passengers with highest fares:")

highest_fares = df.sort_values(by="Fare", ascending=False)

print(highest_fares.head(10))
```

```python
# 2. Calculate the average passenger age.
print("\nAverage passenger age:")

average_age = df["Age"].mean()

print(average_age)
```

```python
# 3. Count the number of male and female passengers.
print("\nNumber of male and female passengers:")

gender_count = df["Sex"].value_counts()

print(gender_count)

```

```python
# 4. Calculate the average fare for each passenger class.
print("\nAverage fare for each passenger class:")

average_fare = df.groupby("Pclass")["Fare"].mean()

print(average_fare)
```

```python
# 5. Calculate the survival rate for each passenger class.
print("\nSurvival rate for each passenger class:")

survival_rate = df.groupby("Pclass")["Survived"].mean()

print(survival_rate)
```
