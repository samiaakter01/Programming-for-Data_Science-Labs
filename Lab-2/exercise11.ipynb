```python
programming_ds = float(input("Enter your Programming for DS mark: "))
data_visualization = float(input("Enter your Data Visualization mark: "))
data_processing = float(input("Enter your Data Processing mark: "))
data_science = float(input("Enter your Data Science mark: "))
java = float(input("Enter your Java mark: "))

marks = [programming_ds, data_visualization, data_processing, data_science, java]

total = 0
maximum = marks[0]
minimum = marks[0]
passed_count = 0
failed_count = 0

for mark in marks:
    total += mark

    if mark > maximum:
        maximum = mark

    if mark < minimum:
        minimum = mark

average_grade = total / len(marks)

print("Total marks:", total)
print("Average grade:", average_grade)
print("Highest mark:", maximum)
print("Lowest mark:", minimum)

for mark in marks:
    if mark >= 50:
        print(mark, "Passed")
        passed_count += 1
    else:
        print(mark, "Failed")
        failed_count += 1

print("Number of students/subjects passed:", passed_count)
print("Number of students/subjects failed:", failed_count)
```

```python
temperatures = (28, 32, 30, 27, 31, 29)

for temperature in temperatures:
    print(temperature)

highest = max(temperatures)
lowest = min(temperatures)

print("Highest temperature:", highest)
print("Lowest temperature:", lowest)
```

```python
text = input("Enter a word or short sentence: ")

characters = len(text)
vowels = 0
spaces = 0

for char in text:
    if char.lower() in "aeiou":
        vowels += 1

    if char == " ":
        spaces += 1

print("Number of characters:", characters)
print("Number of vowels:", vowels)
print("Number of spaces:", spaces)
```

```python
def calculate_grade(marks):
    average = sum(marks) / len(marks)

    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


marks = [85, 72, 90, 78, 88]

grade = calculate_grade(marks)

print("Marks:", marks)
print("Grade:", grade)
```
